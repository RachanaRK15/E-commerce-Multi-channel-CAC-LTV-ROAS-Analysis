```python
import pandas as pd

# 1. Load data
customers = pd.read_csv("customers (1).csv")
transactions = pd.read_csv("transactions (1).csv")
spend = pd.read_csv("marketing_spend (1).csv")

# 2. Clean column names and channel text
for df in (customers, transactions, spend):
  df.columns = df.columns.str.strip().str.lower()

customers["acquisitionchannel"] = (
    customers["acquisitionchannel"].astype(str).str.strip()
)
spend["channel"] = spend["channel"].astype(str).str.strip()

# 3. Remove duplicate rows
customers = customers.drop_duplicates()
transactions = transactions.drop_duplicates()
spend = spend.drop_duplicates()

# 4. Check missing values and convert numbers safely
print(customers.isna().sum())
print(transactions.isna().sum())
print(spend.isna().sum())

transactions["orderamount"] = (
    pd.to_numeric(transactions["orderamount"], errors="coerce").fillna(0)
)
spend["totalspend"] = (
    pd.to_numeric(spend["totalspend"], errors="coerce").fillna(0)
)

# 5. Save cleaned files
customers.to_csv("customers_clean.csv", index=False)
transactions.to_csv("transactions_clean.csv", index=False)
spend.to_csv("marketing_spend_clean.csv", index=False)
```
